## (2) Questions
## Broad Question
-We would like to know which "kinds" of players are most likely to contribute a large amount of data so that we can target those players in our recruiting efforts.
## Specific Question
- Is it possible for age and gender predict total played hours in the Minecraft server dataset?
## Response Variable 
- total_played_hours
## Explanatory Variables
- age and gender
## How the Specific Data Will Help Answer the Question
- The dataset includes the age and gender of the players as well as the total hours spent. We can use this information to see how they are related. Since we have session start and end times we can also obtain the overall playtime. Higher playtime allows for higher contribution of data. 


```r

library(tidyverse)

 players <- read_csv("../players.csv")
sessions <- read_csv("../sessions.csv")
players |>
  summarise(
    mean_age = round(mean(Age, na.rm = TRUE), 2),
    mean_played_hours = round(mean(played_hours, na.rm = TRUE), 2))

sessions <- sessions |>
  mutate(
    start_time = as.numeric(original_start_time),
    end_time = as.numeric(original_end_time),
    duration_minutes = (end_time - start_time) / 60000)

sessions |>
  summarise(
    mean_session_duration_min = round(mean(duration_minutes, na.rm = TRUE), 2))
```

```r
players |> nrow()
sessions |> nrow()
```

```r
players |>
  group_by(gender) |>
  summarize(avg_hours = mean(played_hours, na.rm = TRUE)) |>
  ggplot(aes(x =gender, y= avg_hours, fill = gender)) +
  geom_col() +
  labs(title = "Average Total Played Hours by Gender",
       x = "Gender", y = "Average Played Hours")
```

```r
players |>
  ggplot(aes(x = Age, y = played_hours)) +
  geom_point(color = "steelblue") +
  labs(title = "Age vs Total Played Hours",
       x = "Age",
       y = "Total Played Hours")
```

The bar chart shows the average total hours played by gender.
The scatterplot shows how age relates to total played hours.
We can use both of these to to see how they affect playing time.

Younger players generally spend more time playing. Female and non-binary players have the highest total hours. Players who have longer total time also tend to have longer overall sessions and conribute more data.
