## (1) Data Description

This data was collected from a research group in Computer Science at UBC, led by Frank Wood. He operates two a custom Minecraft server. Each file provided captures different details about the server and the playerbase.
    
## player.csv 
    - contains one row per every unique player.
    - The dataset includes information on whether they have subscribed to the newsletter, region, total time played, and regerstration information.
    - Each player is identifed by a unique ID number.
    - Unit of observation is individual player
    - Expected to have less rows as sessions.csv, because each player can have multiple sessions.
    
## sessions.csv
    - Conatains one row per each gme sessions
    - Records information based on specific sessions such as start time, end time, duration, and in game activity.
    - Each session has a unique session_id and includes the player_id.
    - Unit of observation is individual session.
    - Multiple rows can correspond to the same player.
    
## How the data was collected
    - When a new player joins they receive a unique player and sesssion id.
    - Each time a player logs in to play the server generates a new session id for the player capturing key data with it.
    
## Relationship between the Datasets
    - The two databases are linked by the variable player_id

## Key observations
    - Timezones should be checked to consider timezone consistency.
    - Some players have missing data
    - Outliers with long sessions may exist

## (2) Questions
## Broad Question
- How does age and gender influence how much time the players spend on the minecraft server?
## Specific Question
- Is is possible for age and gender predict total played hours in the Minecraft server dataset?
## Response Variable 
- total_played_hours
## Explanatory Variables
- age and gender
## How the Specific Data Will Help Answer the Question
- The dataset includes the age and gender of the players as well as the total hours spent. We can use this information to see how they arre related. Since we have session start and end times we can also obtain the overall playtime. We can use all this data to create a regression model.
## Data Wrangling Procedure
- Remove any missing values from dataset.
- Convert start and end times into duration
- Group sessions by player to find total playing time

```r
library(tidyverse)
 players <- read_csv("players.csv")
sessions <- read_csv("sessions.csv")

players |> summary()
sessions |> summary()
```

```r
players |> nrow()
sessions |> nrow()
players |> names()
sessions |> names()
```

```r
sessions <- sessions |>
  mutate(
    original_start_time = as.numeric(original_start_time),
    original_end_time   = as.numeric(original_end_time),
    duration = (original_end_time - original_start_time) / 60000)
```

```r
players |>
  ggplot(aes(x = Age)) +
  geom_histogram(binwidth = 2, fill = "skyblue") +
  labs(title = "Distribution of Player Ages",
       x = "Age", y = "Count")

sessions |>
  ggplot(aes(x = duration)) +
  geom_histogram(binwidth = 10, fill = "orange") +
  labs(title = "Distribution of Session Duration",
       x = "Duration (minutes)", y = "Count")
```

There are 1535 recorded sessions from 196 players in the dataset. The player histogram shows most of the players are young, between 15 and 30. This suggests that the server is mostly used by young people. 

The session histogram uses the duratation variable, which measures how long each session was. Most session were relatively short, usally under 30 minutes, however some lasted multiple hours. 

This indicates that while many players log in briefly, a small group of users were on the server for long sessions, this creates variation in playingtimes across the many sessions.

```r
library(tidyverse)

players |>
  ggplot(aes(x = Age, y = played_hours)) +
      geom_point(color = "steelblue") +
  labs(title = "Relationship between Age and Played Hours",
       x = "Age", y = "Played Hours")
players |>
  group_by(gender) |>
  summarize(avg_hours = mean(played_hours, na.rm = TRUE))

player_sessions <- sessions |>
  group_by(hashedEmail) |>
  summarize(avg_duration = mean(duration, na.rm = TRUE)) |>
  left_join(players, by = "hashedEmail")

player_sessions |>
  ggplot(aes(x = played_hours, y = avg_duration)) +
  geom_point(color = "darkorange") +
  labs(title = "Average Session Duration vs Total Played Hours",
       x = "Played Hours", y = "Average Session Duration (minutes)")

```

Younger players generally spend more time playing. Older players tend to player for less time. Female and non-binary players have the highest total hours, with other and prefer not to say having significanlty lowers averages. Players who have longer total time also tend to have longer overall sessions. This patterns suggest that the younger playerbase and specififc genger groups are more active than others and longer total playing times leads to longer sessions.here
