(1) Data Description

This data was collected from a research group in Computer Science at UBC, led by Frank Wood. He operates two a custom Minecraft server. Each file provided captures different details about the server and the playerbase.
    
## player.csv 
    - contains one row per every unique player.
    - The dataset includes information on whether they have subscribed to the newsletter, region, total time played, and regerstration information.
    - Each player is identifed by a unique ID number.
    - Unit of observation is individual player
    - Expected to have less rows as sessions.csv, because each player can have multiple sessions.
    
## sessions.csv
    - Conatains one row per each gme sessions
    - Records information based on specific sessions such as start time, end time, duration, and in game activity.
    - Each session has a unique session_id and includes the player_id.
    - Unit of observation is individual session.
    - Multiple rows can correspond to the same player.
    
## How the data was collected
    - When a new player joins they receive a unique player and sesssion id.
    - Each time a player logs in to play the server generates a new session id for the player capturing key data with it.
    
## Relationship between the Datasets
    - The two databases are linked by the variable player_id

## Key observations
    - Timezones should be checked to consider timezone consistency.
    - Some players have missing data
    - Outliers with long sessions may exist
